```python
from pathlib import Path
import sys
import warnings
warnings.filterwarnings("ignore")

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.utils import (
    RANDOM_STATE, TARGET_COL, TIME_COL, RAW_DIR, PROCESSED_DIR, FIGURES_DIR,
    load_raw_data, set_seed, ensure_dirs, save_figure, PROJECT_ROOT,
)
set_seed(RANDOM_STATE)
ensure_dirs()
print("ROOT =", ROOT)
print("RANDOM_STATE =", RANDOM_STATE)

from src.features import engineer_features
from src.validation import temporal_holdout
from src.modeling import build_logistic_pipeline, fit_and_score
from src.preprocessing import prepare_xy, get_feature_columns

train, _ = load_raw_data()
train = engineer_features(train)
split = temporal_holdout(train, val_ratio=0.20)
Xtr, ytr = prepare_xy(split.train)
Xva, yva = prepare_xy(split.valid)
num, cat = get_feature_columns(split.train)
pipe = build_logistic_pipeline(num, cat)
res = fit_and_score(pipe, Xtr, ytr, Xva, yva, "logistic_regression")
print("F1 @0.5:", round(res["metrics_05"]["f1"],4))
print("F1 @opt:", round(res["metrics_opt"]["f1"],4), "seuil", res["threshold"])
print(res["metrics_opt"]["classification_report"])
```
